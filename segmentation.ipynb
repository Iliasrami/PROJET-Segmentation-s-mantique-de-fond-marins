```python
import os
import cv2
import numpy as np
from detectron2.structures import BoxMode

LABEL_COLORS_LIST = [
    (0, 0, 0),       # background
    (0, 0, 255),     # human
    (0, 255, 0),     # plant
    (0, 255, 255),   # wreck
    (255, 0, 0),     # robot
    (255, 0, 255),   # reef
    (255, 255, 0),   # fish
    (255, 255, 255)  # rocks
]
COLOR_TO_CLASS = {color: idx for idx, color in enumerate(LABEL_COLORS_LIST)}

def bmp_label_to_class_indices(label_path):
    label_image = cv2.imread(label_path, cv2.IMREAD_COLOR)
    label_image = cv2.cvtColor(label_image, cv2.COLOR_BGR2RGB)  # Convert BGR to RGB
    class_indices_image = np.zeros(label_image.shape[:2], dtype=np.uint8)  # Assuming class indices fit in uint8

    for color, class_idx in COLOR_TO_CLASS.items():
        mask = (label_image == color).all(axis=-1)
        class_indices_image[mask] = class_idx

    return class_indices_image

def get_semantic_seg_dicts(img_dir):
    dataset_dicts = []
    for filename in os.listdir(img_dir+ "images/"):
        if not filename.endswith(".jpg"):
            continue

        image_id = os.path.splitext(filename)[0]
        img_path = os.path.join(img_dir+ "images/", filename)
        label_path = os.path.join(img_dir+"train_label/", image_id + ".bmp")

        record = {
            "file_name": img_path,
            "image_id": image_id,
            "height": cv2.imread(img_path).shape[0],
            "width": cv2.imread(img_path).shape[1],
            "sem_seg_file_name": label_path  # Adjusted to include the label path directly
        }

        dataset_dicts.append(record)

    return dataset_dicts
def get_semantic_seg_dicts_test(img_dir):
    dataset_dicts = []
    for filename in os.listdir(img_dir+ "images/"):
        if not filename.endswith(".jpg"):
            continue

        image_id = os.path.splitext(filename)[0]
        img_path = os.path.join(img_dir+ "images/", filename)
        label_path = os.path.join(img_dir+"test_label/", image_id + ".bmp")

        record = {
            "file_name": img_path,
            "image_id": image_id,
            "height": cv2.imread(img_path).shape[0],
            "width": cv2.imread(img_path).shape[1],
            "sem_seg_file_name": label_path  # Adjusted to include the label path directly
        }

        dataset_dicts.append(record)

    return dataset_dicts

# Registering the dataset
from detectron2.data import DatasetCatalog, MetadataCatalog
DatasetCatalog.register("dataset_train", lambda: get_semantic_seg_dicts("data/train_val/"))
MetadataCatalog.get("dataset_train").set(stuff_classes=["background", "human", "plant", "wreck", "robot", "reef", "fish", "rocks"])
DatasetCatalog.register("dataset_test", lambda: get_semantic_seg_dicts_test("data/test/"))
MetadataCatalog.get("dataset_test").set(stuff_classes=["background", "human", "plant", "wreck", "robot", "reef", "fish", "rocks"])
```
